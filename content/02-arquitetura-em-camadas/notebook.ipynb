Neste notebook vamos construir um pequeno gerenciador de tarefas dividido em
três camadas. Depois, vamos quebrar de propósito a regra principal do estilo
para ver o que dá errado.

```
+----------------------------+
|  Apresentação  (tela)      |   recebe comandos e mostra resultados
+-------------+--------------+
              | usa
+-------------v--------------+
|  Domínio  (regras)         |   decide o que é uma tarefa válida
+-------------+--------------+
              | usa
+-------------v--------------+
|  Persistência  (dados)     |   guarda e busca as tarefas
+----------------------------+
```

**Regra principal:** uma camada só pode usar a camada abaixo dela, nunca a de cima.

Roteiro:

1. Seção 2: o sistema implementado do jeito certo.
2. Seção 3: a violação, com a Persistência usando o Domínio.
3. Seção 4: a medição da diferença entre as duas versões.

## 0. Preparação

Usamos só a biblioteca padrão do Python, então não é preciso instalar nada.

````python
import inspect  # usado na seção 4 para ler o código das classes
import sys

print("Versão do Python:", sys.version.split()[0])
````

## 1. O problema

Um gerenciador de tarefas precisa fazer três coisas:

1. Validar os dados: o título não pode ser vazio e a prioridade deve ser `baixa`, `media` ou `alta`.
2. Guardar as tarefas em algum lugar.
3. Mostrar o resultado para quem usa o sistema.

Se tudo isso fica numa classe só, mexer em uma parte (trocar o banco de dados,
criar uma regra nova, trocar o terminal por uma página web) pode quebrar as
outras duas.

A Arquitetura em Camadas separa cada responsabilidade em uma camada e impõe uma
regra: a Apresentação usa o Domínio, o Domínio usa a Persistência, e o caminho
contrário é proibido.

## 2. Implementação seguindo o estilo

### 2.1 Camada de Persistência (`TarefaRepository`)

Só guarda e busca dados. Não valida nada e não sabe que as outras camadas
existem. Aqui as tarefas ficam num dicionário em memória, que faz o papel de um
banco de dados.

Cada tarefa é um dicionário simples, por exemplo:
`{"id": 1, "titulo": "Estudar", "prioridade": "alta", "concluida": False}`.

````python
class TarefaRepository:
    """Camada de PERSISTÊNCIA: guarda e busca tarefas."""

    def __init__(self):
        self.tarefas = {}  # { id: tarefa }
        self.proximo_id = 1

    def salvar(self, tarefa):
        tarefa["id"] = self.proximo_id
        self.tarefas[self.proximo_id] = tarefa
        self.proximo_id = self.proximo_id + 1
        return tarefa

    def buscar_por_id(self, tarefa_id):
        return self.tarefas.get(tarefa_id)  # devolve None se não existir

    def buscar_todas(self):
        return list(self.tarefas.values())

    def atualizar(self, tarefa):
        self.tarefas[tarefa["id"]] = tarefa
````

### 2.2 Camada de Domínio (`TarefaService`)

É onde ficam as regras de negócio. Quando uma regra é desrespeitada, o Domínio
lança um `TarefaInvalidaError`.

O Domínio usa a Persistência, que está abaixo dele, mas não sabe nada sobre a
tela.

````python
class TarefaInvalidaError(Exception):
    """Erro lançado quando uma regra de negócio é desrespeitada."""


class TarefaService:
    """Camada de DOMÍNIO: aplica as regras de negócio."""

    PRIORIDADES_VALIDAS = ["baixa", "media", "alta"]

    def __init__(self, repositorio):
        self.repositorio = repositorio  # usa a camada de baixo: permitido

    def criar_tarefa(self, titulo, prioridade="media"):
        titulo = titulo.strip()

        if titulo == "":
            raise TarefaInvalidaError("O título não pode ser vazio.")

        if prioridade not in self.PRIORIDADES_VALIDAS:
            raise TarefaInvalidaError(
                f"A prioridade '{prioridade}' não existe. Use: baixa, media ou alta."
            )

        nova_tarefa = {"titulo": titulo, "prioridade": prioridade, "concluida": False}
        return self.repositorio.salvar(nova_tarefa)

    def concluir_tarefa(self, tarefa_id):
        tarefa = self.repositorio.buscar_por_id(tarefa_id)

        if tarefa is None:
            raise TarefaInvalidaError(f"A tarefa #{tarefa_id} não existe.")

        tarefa["concluida"] = True
        self.repositorio.atualizar(tarefa)
        return tarefa

    def listar_pendentes(self):
        pendentes = []
        for tarefa in self.repositorio.buscar_todas():
            if not tarefa["concluida"]:
                pendentes.append(tarefa)
        return pendentes
````

### 2.3 Camada de Apresentação (`PresentationCLI`)

Só conversa com o usuário: chama o Domínio e mostra o resultado na tela. Não tem
nenhuma regra de negócio e nunca usa o `TarefaRepository` diretamente.

Se um dia o terminal for trocado por uma página web, só esta classe muda.

````python
class PresentationCLI:
    """Camada de APRESENTAÇÃO: conversa com o usuário pelo terminal."""

    def __init__(self, servico):
        self.servico = servico  # usa a camada de baixo: permitido

    def adicionar(self, titulo, prioridade="media"):
        try:
            tarefa = self.servico.criar_tarefa(titulo, prioridade)
            print(f"Tarefa criada: #{tarefa['id']} {tarefa['titulo']} [{tarefa['prioridade']}]")
        except TarefaInvalidaError as erro:
            print(f"Erro: {erro}")

    def concluir(self, tarefa_id):
        try:
            self.servico.concluir_tarefa(tarefa_id)
            print(f"Tarefa #{tarefa_id} concluída.")
        except TarefaInvalidaError as erro:
            print(f"Erro: {erro}")

    def mostrar_pendentes(self):
        print("--- Tarefas pendentes ---")
        pendentes = self.servico.listar_pendentes()
        if len(pendentes) == 0:
            print("(nenhuma)")
        for tarefa in pendentes:
            print(f"#{tarefa['id']} [{tarefa['prioridade']}] {tarefa['titulo']}")
````

### 2.4 Montando e usando o sistema

As camadas são criadas de baixo para cima, e cada uma recebe a camada que está
abaixo dela.

Repare que três chamadas são recusadas. Quem recusa é o Domínio, e a
Apresentação só mostra a mensagem de erro.

````python
repositorio = TarefaRepository()
servico = TarefaService(repositorio)
tela = PresentationCLI(servico)

tela.adicionar("Estudar arquitetura em camadas", "alta")
tela.adicionar("Revisar slides de HCI", "media")
tela.adicionar("", "alta")             # título vazio: recusado
tela.adicionar("Tarefa X", "urgente")  # prioridade inválida: recusado

tela.concluir(1)
tela.concluir(99)                      # tarefa inexistente: recusado

print()
tela.mostrar_pendentes()
````

### 2.5 Cada camada pode ser testada sozinha

Como a Persistência não depende de ninguém, ela pode ser testada sem criar o
Domínio nem a tela. Guarde isso, porque é justamente o que a violação vai
quebrar.

````python
# Testando só a Persistência: nenhuma outra camada é criada
repositorio_de_teste = TarefaRepository()
repositorio_de_teste.salvar({"titulo": "Teste", "prioridade": "baixa", "concluida": False})

assert len(repositorio_de_teste.buscar_todas()) == 1
print("Teste passou: a Persistência funcionou sem o Domínio.")
````

## 3. A violação

Imagine um requisito novo: "toda vez que uma tarefa for salva, o Domínio deve
ser avisado". A solução ingênua é entregar o `TarefaService` ao repositório e
fazer o repositório chamá-lo:

```
Persistência --usa--> Domínio      (proibido: aponta para cima)
```

O código abaixo faz exatamente isso. Ele roda sem erro, mas os problemas
aparecem logo em seguida.

````python
# ANTIPADRÃO: a Persistência passa a usar o Domínio (dependência para cima)

class TarefaRepositoryViolador(TarefaRepository):
    """Versão ERRADA da Persistência: conhece e chama o Domínio."""

    def __init__(self, servico):
        super().__init__()
        self.servico = servico  # a Persistência conhecendo o Domínio

    def salvar(self, tarefa):
        tarefa_salva = super().salvar(tarefa)
        self.servico.listar_pendentes()  # chamando a camada de cima
        return tarefa_salva
````

### 3.1 Primeiro problema: um precisa do outro para existir

O `TarefaService` precisa de um repositório para ser criado, e o
`TarefaRepositoryViolador` precisa de um serviço para ser criado. É um ciclo de
dependências.

A única saída é uma gambiarra: criar um deles incompleto e consertá-lo depois.

````python
# 1. Cria o serviço ainda sem repositório
servico_violador = TarefaService(repositorio=None)

# 2. Cria o repositório, que exige o serviço
repositorio_violador = TarefaRepositoryViolador(servico_violador)

# 3. Volta no serviço e coloca o repositório à mão
servico_violador.repositorio = repositorio_violador

servico_violador.criar_tarefa("Funciona, mas só com gambiarra", "media")
print("Funcionou, mas só porque ligamos as duas classes uma na outra à mão.")
````

### 3.2 Segundo problema: a Persistência não pode mais ser testada sozinha

Vamos repetir o teste da seção 2.5, agora com o repositório violador e sem
nenhum Domínio por perto.

````python
repositorio_isolado = TarefaRepositoryViolador(servico=None)

try:
    repositorio_isolado.salvar({"titulo": "Teste", "prioridade": "baixa", "concluida": False})
    print("Teste passou: a Persistência funcionou sem o Domínio.")
except AttributeError as erro:
    print("Teste falhou: a Persistência não funciona mais sem o Domínio.")
    print("Erro do Python:", erro)
````

## 4. Consequência mensurável

Em vez de só afirmar que a versão violadora é pior, vamos medir. Para cada
classe de Persistência, verificamos duas coisas:

1. Citações ao Domínio: lemos o código de cada método da classe e contamos
   quantas vezes aparece a palavra `servico`. Uma Persistência correta deve ter zero.
2. Testável sozinha: tentamos salvar uma tarefa sem nenhum Domínio por perto.

````python
def contar_citacoes_ao_dominio(classe):
    """Conta quantas vezes a palavra 'servico' aparece nos métodos da classe."""
    total = 0
    for metodo in vars(classe).values():
        if callable(metodo):
            codigo = inspect.getsource(metodo)
            total = total + codigo.count("servico")
    return total


def funciona_sem_dominio(repositorio):
    """Tenta salvar uma tarefa. Devolve True se der certo."""
    try:
        repositorio.salvar({"titulo": "Teste", "prioridade": "baixa", "concluida": False})
        return True
    except AttributeError:
        return False


repositorios = [TarefaRepository(), TarefaRepositoryViolador(servico=None)]

print(f"{'Classe':28} | {'Citações ao Domínio':20} | Testável sozinha?")
print("-" * 70)
for repositorio in repositorios:
    classe = type(repositorio)
    citacoes = contar_citacoes_ao_dominio(classe)
    testavel = "sim" if funciona_sem_dominio(repositorio) else "não"
    print(f"{classe.__name__:28} | {citacoes:<20} | {testavel}")
````

## 5. Conclusão

A tabela resume a diferença:

- `TarefaRepository` não cita o Domínio e pode ser testado sozinho. Dá para
  trocá-lo por um banco de dados real sem mexer no resto do sistema.
- `TarefaRepositoryViolador` cita o Domínio, só pode ser criado com gambiarra e
  deixa de funcionar sozinho. Persistência e Domínio passam a ser uma peça só.

A regra "as dependências só apontam para baixo" é o que garante que cada camada
possa ser testada, trocada e evoluir sem arrastar as outras junto.
